# Compile network_diagram.tex with pdflatex

This notebook compiles the TikZ figure in `src/plots/network_diagram.tex` and writes the output PDF next to the source file.

```python
from pathlib import Path
import subprocess
import sys

tex_file = Path('network_diagram.tex')

if not tex_file.exists():
    candidates = [
        Path('src/plots/network_diagram.tex'),
        Path('plots/network_diagram.tex'),
        Path('network_diagram.tex'),
    ]
    for c in candidates:
        if c.exists():
            tex_file = c
            break
    else:
        raise FileNotFoundError('Could not find network_diagram.tex. Check the current working directory.')

tex_dir = tex_file.parent
tex_name = tex_file.name
pdf_path = tex_dir / (tex_file.stem + '.pdf')

cmd = [
    'pdflatex',
    '-interaction=nonstopmode',
    '-halt-on-error',
    '-output-directory',
    str(tex_dir),
    str(tex_name),
]

print(f'Compiling: {tex_file}')
print('Running command:', ' '.join(cmd))

result = subprocess.run(cmd, cwd=str(tex_dir), capture_output=True, text=True)

if result.stdout:
    print(result.stdout)
if result.stderr:
    print(result.stderr)

if result.returncode != 0:
    raise SystemExit(f'pdflatex failed with exit code {result.returncode}')

print(f'PDF generated at: {pdf_path}')
print('Done.')
```
