# Channel Model Notebook

This notebook loads the channel propagation model from `channel_model.py` and runs a few representative calculations and plots used in the HAPS project.

- HAPS FSPL
- UAV/ground path loss
- HAPS SINR and multi-node interference
- example plot generation

```python
from pathlib import Path
import sys

# Make the workspace and model directory importable from the notebook
root = Path.cwd().resolve()
candidates = [
    root,
    root.parent,
    root / 'src',
    root / 'src' / 'models',
]
for candidate in candidates:
    if candidate.exists():
        sys.path.insert(0, str(candidate))

print('Notebook working directory:', root)
print('Python path configured for local imports.')
```

```python
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

from channel_model import (
    ENVIRONMENTS,
    haps_fspl_db,
    haps_a2g_pathloss_db,
    uav_a2g_pathloss_db,
    multi_node_sinr_db,
    rx_power_dbm,
    make_plots,
)

print('Imported channel_model symbols successfully.')
```

```python
# Example 1: HAPS free-space path loss for a few distances
distances_km = np.array([1, 10, 50, 100])
fspl_db = haps_fspl_db(2000.0, distances_km)
print('HAPS FSPL [dB]:')
for d, val in zip(distances_km, fspl_db):
    print(f'{d:>5.0f} km -> {val:>7.2f} dB')
```

```python
# Example 2: UAV average path loss vs distance for one environment
r = np.linspace(1, 1500, 200)
env = 'urban'
pl = uav_a2g_pathloss_db(r, h_uav_m=300.0, env=env)
plt.figure(figsize=(8, 5))
plt.plot(r, pl, linewidth=2)
plt.title('UAV air-to-ground path loss (300 m altitude)')
plt.xlabel('Horizontal distance [m]')
plt.ylabel('Average path loss [dB]')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Example 3: Multi-node HAPS SINR over a range of distances
d_m = np.linspace(1e3, 100e3, 200)
sinr_db = multi_node_sinr_db(d_m, num_haps=3, h_node_m=20000.0, env='urban')
plt.figure(figsize=(8, 5))
plt.plot(d_m / 1000.0, sinr_db, linewidth=2, color='tab:blue')
plt.axhline(0, color='gray', linestyle='--', linewidth=1)
plt.title('HAPS SINR with multi-node interference')
plt.xlabel('Horizontal distance [km]')
plt.ylabel('SINR [dB]')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
